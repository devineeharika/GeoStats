```
import pandas as pd
import numpy
df=pd.read_csv('/content/Soil_Nutrients.csv', sep=',',header=None)
list_data = df.to_numpy().tolist()
print(list_data)
print(type(list_data))
long_data=[]
for i in range(len(list_data)):
  long_data.append([])
  for k in range(1,3):
    long_data[i].append(list_data[i][k])
print(long_data)
p_data=[]
for i in range(len(list_data)):
  p_data.append(list_data[i][7])
print(p_data)
k_data=[]
for i in range(len(list_data)):
  k_data.append(list_data[i][8])
print(k_data)
s_data=[]
for i in range(len(list_data)):
  s_data.append(list_data[i][9])
print(s_data)
zn_data=[]
for i in range(len(list_data)):
  zn_data.append(list_data[i][10])
print(zn_data)
fe_data=[]
for i in range(len(list_data)):
  fe_data.append(list_data[i][11])
print(fe_data)
cu_data=[]
for i in range(len(list_data)):
  cu_data.append(list_data[i][12])
print(cu_data)
mn_data=[]
for i in range(len(list_data)):
  mn_data.append(list_data[i][13])
print(mn_data)
b_data=[]
for i in range(len(list_data)):
  b_data.append(list_data[i][14])
print(b_data)
```

```
import numpy as np
import skgstat as skg

coordinates = long_data
values_1 =p_data
values_2=k_data
values_3=s_data
values_4=zn_data
values_5=fe_data
values_6=cu_data
values_7=mn_data
values_8=b_data
V1= skg.Variogram(coordinates, values_1,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V2= skg.Variogram(coordinates, values_2,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V3= skg.Variogram(coordinates, values_3,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)  
V4= skg.Variogram(coordinates, values_4,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V5= skg.Variogram(coordinates, values_5,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V6= skg.Variogram(coordinates, values_6,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V7= skg.Variogram(coordinates, values_7,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
V8= skg.Variogram(coordinates, values_8,normalize=False, n_lags=20, maxlag=np.max(dists)/2.0)
print(V1)
V1.plot()
print(V2)
V2.plot()
print(V3)
V3.plot()
print(V4)
V4.plot()
print(V5)
V5.plot()
print(V6)
V6.plot()
print(V7)
V7.plot()
print(V8)
V8.plot()
```
