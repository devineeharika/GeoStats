```
import pandas as pd
import numpy
df=pd.read_csv('/content/phoscaled.csv', sep=',',header=None)
list_data = df.to_numpy().tolist()
print(list_data)
print(type(list_data))
long_data=[]
for i in range(len(list_data)):
  long_data.append([])
  for k in range(2):
    long_data[i].append(list_data[i][k])
print(long_data)
P_data=[]
for i in range(len(list_data)):
  P_data.append(list_data[i][2])
print(P_data)
```

```
!pip install scikit-gstat
```

```

import numpy as np
import skgstat as skg

coordinates = long_data
values =P_data
V = skg.Variogram(coordinates, values,normalize=False, n_lags=20, maxlag=0.5 , bin_func = 'uniform' , verbose = True ,use_nugget = True)
   
print(V)
V.plot()
```

```
import pandas as pd
import numpy
df=pd.read_csv('/content/Pscaled.csv', sep=',',header=None)
list_data = df.to_numpy().tolist()
long_data=[]
for i in range(len(list_data)):
     long_data.append(list_data[i][1])
print(long_data)
P_data=[]
for i in range(len(list_data)):
  P_data.append(list_data[i][2])
print(P_data)
lat_data=[]
for i in range(len(list_data)):
  lat_data.append(list_data[i][0])
print(lat_data)

print(len(long_data))
```

```
import pandas as pd
import numpy
df1=pd.read_csv('/content/gridxy.csv', sep=',',header=None)
test_data = df1.to_numpy().tolist()
longtest_data=[]
for i in range(len(test_data)):
     longtest_data.append(test_data[i][1])
print(longtest_data)

lattest_data=[]
for i in range(len(test_data)):
  lattest_data.append(test_data[i][0])
print(lattest_data)
print(len(lattest_data))
```

```
!pip install pykrige
```

```
from pykrige.ok import OrdinaryKriging
import numpy as np
import matplotlib.pyplot as plt
import time 

start = time.time()
# Create ordinary kriging object:
OK = OrdinaryKriging(
    long_data[:23500],
    lat_data[:23500],
    P_data[:23500],
    variogram_model="spherical",
    variogram_parameters = {'sill' : 0.01 , 'nugget' : 0.00, 'range' :1.85 },
    verbose=True,
    enable_plotting=True,
    coordinates_type="geographic",
    nlags=20,
)

# Execute on grid:
z1, ss1 = OK.execute("points", longtest_data, lattest_data)

# # Create ordinary kriging object ignoring curvature:
# OK = OrdinaryKriging(
    
#     long_data[:23500],
#      lat_data[:23500],
#       N_data[:23500], 
#       variogram_model="spherical",
#       variogram_parameters = {'sill' : 0.02 , 'nugget' : 0.01, 'range' :1.27 },
#       nlags=20, 
#       verbose=True,
#        enable_plotting=True
# )l=

# # Execute on grid:
# z2, ss2 = OK.execute("points", longtest_data[:10], lattest_data[:10])

# Print data at equator (last longitude index will show periodicity):
print("Original data:")
print("Longitude:", np.array(long_data  ,dtype = np.float32))
print("Latitude: ", np.array(lat_data  ,dtype = np.float32))
print("z:        ", np.array(P_data))
print("\nKrige values:\n======================")
print("Longitude:", longtest_data[:10])
print("Value:    ", np.array(z1))
print("Sigma²:   ", np.array(ss1))
# print("\nIgnoring curvature:\n=====================")
# print("Value:    ", np.array(z2))
# print("Sigma²:   ", np.array(ss2))
print('time taken : ' , time.time() - start)

plt.scatter(lattest_data , longtest_data , c = z1 , cmap = 'gist_rainbow')
cbar= plt.colorbar()
cbar.set_label("Concentration (m)", labelpad=+1)
plt.xlabel("Latitude")
plt.ylabel("Longitude")
plt.title("Predicted data")
plt.show()


# plt.imshow(z1, origin="lower")
# plt.show()
# plt.imshow(z2, origin="lower")
# plt.show()
```
